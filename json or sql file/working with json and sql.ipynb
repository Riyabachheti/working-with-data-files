# Working with SQL

To work with SQL data, we'll need a local server environment. For this, we're using XAMPP, a free, open-source package that includes the Apache web server and the MySQL database. We need XAMPP to run our SQL file on a database on our local machines.

-----

## How We Work with SQL Data

Here are the steps we follow to get our SQL data set up and running:

1.  **Download and Install XAMPP**: First, we download and install XAMPP. This provides us with the necessary **Apache** and **MySQL** servers.

2.  **Start Services**: We open the XAMPP control panel and start both the **Apache** and **MySQL** services.

3.  **Access phpMyAdmin**: In our web browser, we navigate to `localhost/phpmyadmin`.

4.  **Create a New Database**: In the phpMyAdmin interface, we click on "New" to create a new database. We'll name it "world" and click "Create".

5.  **Import SQL File**: We select the "world" database we just created and go to the "Import" tab. We choose our `.sql` file and click "Go" to import the data.

6.  **Install MySQL Connector**: To connect Python with our MySQL database, we need to install the `mysql.connector` library. We do this using pip in our terminal:

    ```bash
    pip install mysql.connector
    ```

7.  **Connect in Python**: In our Python script, we import the library and establish a connection to our database with this code:

    ```python
    import mysql.connector

    conn = mysql.connector.connect(
        host="localhost",
        user="root",
        password="",
        database="world"
    )
    ```

8.  **Execute SQL Queries**: Now we can execute SQL queries using the pandas library. The `pd.read_sql_query()` function lets us run a query and load the results directly into a DataFrame, which is perfect for our data analysis needs\! 🐼

## 1. Direct MySQL Connector Approach
#### Connects directly to the MySQL database using its specific driver (mysql.connector).
#### Works, but raises a UserWarning. It's not the officially recommended or tested method.
#### Creates a basic connection object. Your connection code is tied to mysql.connector. Switching to another database (e.g., PostgreSQL) requires rewriting the connection logic.
#### Simpler, but less flexible for production code.


````python
!pip install mysql.connector
````

````python
import mysql.connector
````

````python
conn = mysql.connector.connect(host='localhost',user='root',password='',database='world')
````

````python
pd.read_sql_query("SELECT * FROM countrylanguage WHERE CountryCode = 'ABW'",conn)
````

## 2. SQLAlchemy + Pandas Approach
#### Uses SQLAlchemy as a middleman. SQLAlchemy then uses a driver (like mysql-connector-python) to talk to the database.
#### Officially Recommended. This is the standard, warning-free way to connect pandas to SQL databases.
#### Highly flexible. To switch to PostgreSQL, you just change the connection string. Your pandas code remains identical.
#### Creates a more powerful engine object, which can manage a pool of connections for better performance in larger applications.
#### Most applications, especially when using pandas. It's more robust, flexible, and future-proof.

````python
!pip install SQLAlchemy mysql-connector-python
````

````python
from sqlalchemy import create_engine

# 1. Create a database engine using SQLAlchemy
# The format is: 'mysql+mysqlconnector://user:password@host/database_name'
engine = create_engine("mysql+mysqlconnector://root:@localhost/world")

# 2. Use the engine with pandas (this will not produce a warning)
df = pd.read_sql_query("SELECT * FROM countrylanguage", engine)

# Print the first 5 rows
print(df.head())
````
