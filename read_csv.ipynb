# 3. opening a csv file from a url

## <font color='red'> We can use the following code to get data of csv file from any server.</font>
We're using requests module.
Here our program act as a user/client and request the data. This request is sent with the header defined in "headers" below.
we fetch all the data of file from a url using requests.get() and store it in req variable.


```python
import requests
from io import StringIO
import pandas as pd

url = "https://raw.githubusercontent.com/datasets/covid-19/master/data/countries-aggregated.csv"
headers = {
    "User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"
}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
```

# Steps to Get the Raw CSV Link from a github repo

1. Go to the repository on GitHub where the .csv file is located.

    Example: https://github.com/cs109/2014_data

2. Navigate inside the repo until you see the .csv file (e.g., olympicAthletes.csv).

3. Click on the file name → GitHub will open a preview of the CSV.

4. At the top-right, you’ll see a button “Raw”.

5. Click it → it opens the file in raw text format.

6. Copy the URL from your browser.
    Here url will look like this : https://github.com/cs109/2014_data/blob/master/olympicAthletes.csv

8.Now, Replace github.com → raw.githubusercontent.com and Remove the word blob/ from the path

   here url for raw csv: https://raw.githubusercontent.com/cs109/2014_data/master/olympicAthletes.csv

```python
import requests
from io import StringIO
url = "https://raw.githubusercontent.com/thewiremonkey/factbook.csv/master/categories.csv"
headers = {
    "User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"
}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
```

## 11. Skip bad lines

```python
import requests
from io import StringIO
import pandas as pd

url = "https://raw.githubusercontent.com/cs109/2014_data/master/olympicAthletes.csv"

headers = {"User-Agent": "Mozilla/5.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

df = pd.read_csv(data,sep='\t',names=['id','athelete name','gender','age','sport'],index_col='id',skiprows=1, on_bad_lines="skip")   # skips problematic rows
print(df.head())
```

```python
df= pd.read_csv('BX_Books.csv', sep=';', encoding="latin-1", on_bad_lines='skip', nrows=3)
df
```

## 13. Handling dates
Most of the times the date column is passed as string in databases and we cannot perform datetime operations on them, So we parse the date column so we can use it's functionalities properly.

```python
pd.read_csv('IPL Matches 2008-2020.csv').info()
```

```python
df = pd.read_csv('IPL Matches 2008-2020.csv', parse_dates=['date'])
df.info()
df.head()
```

## 14. Converters

```python
# Function definition used for converters/transformation of column.
def rename(name):
    if name == "Royal Challengers Bangalore":
        return "RCB"
    else:
        return name

# Calling the function
rename("Royal Challengers Bangalore")

# Using the function as a converter
pd.read_csv('IPL Matches 2008-2020.csv', converters={'team1':rename}).head()
```

## 15. na_value parameter
Sometimes the NaN values are not very direct, they are in forms of hiphen (-) or something else. In these cases we can specify the kind of values we are considering NaN

```python
pd.read_csv('aug_train.csv',na_values = ['no_enrollment', 'never']).head()
# in 'enrolled_university' column all the values 'no_enrollment' are converted to NaN. Similarly in 'last_new_job' column all the 'never' values are converted to NaN
```
