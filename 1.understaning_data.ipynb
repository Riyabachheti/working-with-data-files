#### Data Analysis 
Before we start working on our data we try to understand our dataset first, for this we ask some basic questions. These initial questions provide a solid foundation for understanding your dataset before proceeding with more advanced EDA and machine learning model building.

The core idea is to ask seven basic questions about the data to understand it better.
Here are the questions and how to answer them using Python's Pandas library:

dataset used : titatic 
#### How big is our data?
**Purpose:** To understand the dimensions (rows and columns) of our dataset. This helps in anticipating computational requirements and understanding the scale of the data.

**Method:** Use df.shape.

Example: For the Titanic dataset, df.shape would return something like (891, 12), indicating 891 rows and 12 columns. Knowing this beforehand is crucial, especially when dealing with very large datasets.

```python
import pandas as pd
```

```python
df = pd.read_csv('train.csv')
```

# 1. How big is the data?

```python
df.shape
```

# 2. How does the data look like?
(First few rows, random rows)


**Purpose:** To get a quick visual overview of the data's structure and content.  

**Methods:**  
- `df.head()`: Shows the top 5 rows of the dataset.  
- `df.sample(5)`: Displays 5 random rows from the dataset.  

<span style="color:blue">This is a better approach than `df.head()` because it helps avoid bias.</span>  
If the initial rows are homogenous (e.g., all children's data), `head()` might give a misleading impression of the entire dataset. Sampling randomly provides a more generalized view.  


```python
df.head()
```

```python
df.sample(5)
```

# 3. What is the data type of cols?

**Purpose:** To identify the data type (e.g., integer, float, object/string) of each column, which is critical for future operations and memory optimization.  

**Method:** Use `df.info()`.  

**Output:** This function provides extensive information:  
- Number of non-null values for each column.  
- Data type (dtype) of each column (e.g., int64, float64, object).  
- Memory usage of the DataFrame.  

<span style="color:darkred">**Memory Optimization Insight:**</span> Sometimes, numerical columns might be stored as float types unnecessarily (e.g., an ID column stored as float64 when int64 would suffice). Converting these to appropriate integer types can reduce memory consumption and improve algorithm speed, especially for large datasets.  


```python
df.info()
```

# 4. Are there any missing values?


**Purpose:** Missing values (NaNs) can significantly impact model performance and require careful handling (imputation or removal). It's crucial to know their presence and quantity upfront.  

**Observation from `df.info()`:** `df.info()` already shows the count of non-null values. If this count is less than the total number of rows, there are missing values.  

**Method for exact counts:** Use `df.isnull().sum()`.  

**Output:** This returns a Series showing the number of missing values for each column.  

**Actionable Insight:** Columns with a high number of missing values (e.g., **"Cabin" with 687 missing values out of 891**, or **"Age" with 177**) require a decision: either fill the missing values (imputation) or remove the entire column.  


```python
df.isnull().sum()
```

# 5. How does the data look mathematically?

**Purpose:** To obtain a high-level mathematical summary of our numerical columns. This helps in understanding the distribution and identifying potential outliers or anomalies.  

**Method:** Use `df.describe()`.  

**Output:** Provides key statistical metrics for numerical columns:  
- **count:** Number of non-null observations.  
- **mean:** Average value.  
- **std:** Standard deviation.  
- **min:** Minimum value.  
- **25% (Q1):** First quartile (25th percentile).  
- **50% (median):** Median (50th percentile).  
- **75% (Q3):** Third quartile (75th percentile).  
- **max:** Maximum value.  

**Insights from Titanic Dataset:**  
- **Age:** Mean age is around **30**, standard deviation is **14**. Minimum age is **0.42** (infant), median is **28**, and 75% of passengers are **38 or younger**.  
  <span style="color:darkred">This helps detect anomalies if, for instance, a minimum age was 200.</span>  
- **Fare:** Can provide insights into ticket pricing and distribution.  

**Limitation:** `df.describe()` only works on numerical columns.  


```python
df.describe()
```

# 6. Are there duplicate values?

**Purpose:** Duplicate rows can lead to skewed results and an overrepresentation of certain data points in our algorithms. Identifying and removing them is crucial for data integrity.  

**Method:** Use `df.duplicated().sum()`.  

**Output:** Returns the count of duplicate rows. For the Titanic dataset, it shows **0**, meaning all rows are unique.  
If duplicates exist, we can use `df.drop_duplicates()` to remove them.  


```python
df.duplicated().sum()
```

# 7. How is the correlation between cols?

**Purpose:** To understand the relationship between different columns, especially between input features and the target variable (output). This helps in feature selection, as highly correlated or uncorrelated features might be removed.  

**Method:** Use `df.corr()`.  

**Output:** Produces a correlation matrix showing the **Pearson correlation coefficient** between all pairs of numerical columns.  
The values range from **-1 to +1**.  
- **+1:** Strong positive correlation (as one increases, the other increases).  
- **-1:** Strong negative correlation (as one increases, the other decreases).  
- **0:** No linear correlation.  

**Insights from Titanic Dataset (Target variable: Survived):**  
- **Fare and Survived:** Positive correlation (~0.27). <span style="color:green">As Fare increases, the chance of survival tends to increase.</span> This aligns with the understanding that wealthier passengers (paying higher fares) might have had better access to lifeboats.  
- **Pclass and Survived:** Strong negative correlation (~-0.34). <span style="color:darkred">As Pclass increases (meaning cheaper tickets, e.g., Pclass 3), the chance of survival decreases.</span> This is also historically accurate, as lower-class passengers had higher mortality rates.  
- **PassengerId and Survived:** Very low correlation (~0.005). This indicates PassengerId is not a useful feature for predicting survival and can likely be dropped.  

**Limitation:** Like `describe()`, `corr()` also works primarily on numerical columns.  


⚠️ **Note on Pandas versions:**  
- In older versions (≤ 1.4.x), `df.corr()` automatically ignored non-numeric columns.  
- In newer versions (≥ 1.5.0), you must explicitly specify numeric columns; otherwise, Pandas will attempt to convert string/object columns (e.g., `"Name"`) to floats and raise an error.  


```python
df.corr(numeric_only=True)
# khud khudse correlation is 1
```
